```python
import pandas as pd

url = "https://raw.githubusercontent.com/namekonnen/portfolio-optimizer/main/prices.csv"
prices = pd.read_csv(url, index_col="Date", parse_dates=True)

prices.tail(3)
```

```python
from pypfopt import expected_returns

mu = expected_returns.mean_historical_return(prices)

mu
```

```python
from pypfopt import risk_models

S = risk_models.sample_cov(prices)

S.round(3)
```

```python
from pypfopt import EfficientFrontier

ef = EfficientFrontier(mu, S)

weights = ef.max_sharpe(risk_free_rate=0.04)

ef.clean_weights()
```

```python
ef.portfolio_performance(verbose=True, risk_free_rate=0.04)
```

```python
ef_min = EfficientFrontier(mu, S)

ef_min.min_volatility()

print(ef_min.clean_weights())
ef_min.portfolio_performance(verbose=True, risk_free_rate=0.04)
```
